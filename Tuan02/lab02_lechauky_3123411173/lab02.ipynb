# Bài 6. Phép toán và nhân ma trận
Thực hiện các phép toán phần tử và phép nhân ma trận bằng toán tử @.

```python
import torch

a = torch.randn(2, 3)
b = torch.randn(2, 3)

print("a:")
print(a)
print("b:")
print(b)

print("\na + b:")
print(a + b)

print("\na / b:")
print(a / b)

print("\na^2:")
print(a ** 2)

print("\nMatrix multiplication:")
print(a @ b.T)
```

**Yêu cầu:** Phân biệt phép toán a * b với a @ b.T.

```python
print("a * b (nhân từng phần tử):")
print(a * b)
print("Shape:", (a * b).shape)

print("\na @ b.T (nhân ma trận):")
print(a @ b.T)

# Phần tử [0, 0] của a @ b.T bằng tổng hàng 0 của a * b
print("\nKiểm tra: (a @ b.T)[0, 0] =", (a @ b.T)[0, 0].item(),
      "| (a * b)[0].sum() =", (a * b)[0].sum().item())
```

**Yêu cầu:** Cho biết shape của kết quả a @ b.T.

```python
print("a.shape   =", a.shape)
print("b.T.shape =", b.T.shape)
print("(a @ b.T).shape =", (a @ b.T).shape)
```

# Bài 7. Các hàm thống kê trên Tensor
Tính trung bình, độ lệch chuẩn và tổng tích lũy theo từng cột.

```python
import torch

a = torch.randn(3, 4)

print(a)
print("\nMean:")
print(torch.mean(a, dim=0))

print("\nStandard deviation:")
print(torch.std(a, dim=0))

print("\nCumulative sum:")
print(torch.cumsum(a, dim=0))
```

**Yêu cầu:** Thay dim=0 bằng dim=1 và nhận xét sự khác nhau.

```python
print("Mean (dim=1):")
print(torch.mean(a, dim=1))

print("\nStandard deviation (dim=1):")
print(torch.std(a, dim=1))

print("\nCumulative sum (dim=1):")
print(torch.cumsum(a, dim=1))

print("\nShape mean dim=0:", torch.mean(a, dim=0).shape,
      "| Shape mean dim=1:", torch.mean(a, dim=1).shape)
```

# Bài 8. Autograd cơ bản
Tạo biến có requires_grad=True và dùng backward() để tính đạo hàm tự động.

```python
import torch

x = torch.tensor(
    3.6,
    requires_grad=True
)

y = x * x

y.backward()

print("x =", x)
print("y =", y)
print("x.grad =", x.grad)
```

**Yêu cầu:** Tính thủ công đạo hàm của y = x² tại x = 3.6 và so sánh với x.grad.

```python
manual = 2 * 3.6
print("Đạo hàm thủ công: dy/dx = 2x =", manual)
print("x.grad (autograd):", x.grad.item())
print("Trùng khớp:", abs(manual - x.grad.item()) < 1e-5)
```

**Yêu cầu:** Đổi hàm thành y = 3*x**2 + 2*x + 1 và kiểm tra gradient.

```python
x = torch.tensor(3.6, requires_grad=True)
y = 3 * x**2 + 2 * x + 1
y.backward()

manual = 6 * 3.6 + 2
print("y =", y.item())
print("x.grad (autograd):", x.grad.item())
print("Thủ công: dy/dx = 6x + 2 =", manual)
print("Trùng khớp:", abs(manual - x.grad.item()) < 1e-4)
```

# Bài 9. Gradient Descent tìm hệ số đa thức
Sinh dữ liệu từ đa thức y = x² + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.

```python
import numpy as np
import torch

polynomial = np.poly1d([1, 2, 3])
N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

for epoch in range(1000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("Learned coefficients:")
print(w)
```

**Yêu cầu:** So sánh ba hệ số học được với [1, 2, 3].

```python
true_w = torch.tensor([1.0, 2.0, 3.0])
learned = w.detach().ravel()
print("Hệ số thật    :", true_w.tolist())
print("Hệ số học được:", [round(v, 4) for v in learned.tolist()])
print("Sai lệch tuyệt đối:", [round(v, 4) for v in (learned - true_w).abs().tolist()])
print("MSE cuối cùng:", mse.item())
```

**Yêu cầu:** Giải thích vai trò của optimizer.zero_grad(), mse.backward() và optimizer.step().

# Bài 10. Giải hệ phương trình bằng Autograd
Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.

```python
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)

EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2

    sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

**Yêu cầu:** Kiểm tra nghiệm thu được với bốn phương trình: A+B=9, C-D=1, A+C=8, B-D=2.

```python
with torch.no_grad():
    print("A + B =", round((A + B).item(), 4), " (cần = 9)")
    print("C - D =", round((C - D).item(), 4), " (cần = 1)")
    print("A + C =", round((A + C).item(), 4), " (cần = 8)")
    print("B - D =", round((B - D).item(), 4), " (cần = 2)")
    print("Tổng sai số bình phương:", sqerr.item())
```

**Yêu cầu:** Nhận xét vì sao bài toán có thể có nhiều nghiệm.

```python
# Hạng của ma trận hệ số (các cột ứng với A, B, C, D)
M = torch.tensor([[1., 1., 0., 0.],    # A + B = 9
                  [0., 0., 1., -1.],   # C - D = 1
                  [1., 0., 1., 0.],    # A + C = 8
                  [0., 1., 0., -1.]])  # B - D = 2
print("Số ẩn:", M.shape[1], "| Hạng ma trận hệ số:", torch.linalg.matrix_rank(M).item())

# Họ nghiệm tổng quát: D = t, C = 1 + t, B = 2 + t, A = 7 - t
for t in [0.0, 1.0, 2.5]:
    a_, b_, c_, d_ = 7 - t, 2 + t, 1 + t, t
    print(f"t = {t}: A = {a_}, B = {b_}, C = {c_}, D = {d_} ->",
          a_ + b_, c_ - d_, a_ + c_, b_ - d_)
```

## d) Thay đa thức trong Bài 9 bằng y = 2x² - 3x + 5; huấn luyện và cho biết các hệ số tìm được.

```python
import numpy as np
import torch

np.random.seed(0)
torch.manual_seed(0)

polynomial = np.poly1d([2, -3, 5])   # y = 2x^2 - 3x + 5
N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)


def train(lr, epochs):
    w = torch.randn(3, 1, requires_grad=True)
    optimizer = torch.optim.NAdam([w], lr=lr)
    for epoch in range(epochs):
        optimizer.zero_grad()
        y_pred = x @ w
        mse = torch.mean(torch.square(y - y_pred))
        mse.backward()
        optimizer.step()
    return w.detach().ravel(), mse.item()


# Lần 1: giữ nguyên tham số như Bài 9 (lr = 0.01, 1000 epoch)
w1, mse1 = train(lr=0.01, epochs=1000)
print("lr = 0.01, 1000 epoch:")
print("  Hệ số tìm được:", [round(v, 4) for v in w1.tolist()])
print("  MSE:", round(mse1, 6))

# Lần 2: tăng learning rate và số epoch để hội tụ tốt hơn
w2, mse2 = train(lr=0.1, epochs=3000)
print("\nlr = 0.1, 3000 epoch:")
print("  Hệ số tìm được:", [round(v, 4) for v in w2.tolist()])
print("  MSE:", mse2)

print("\nHệ số thật: [2, -3, 5]")
```
